```python
# 1. 데이터 프레임 읽어오기
import pandas as pd
weather = pd.read_excel("./assets/weather2023.xlsx")
weather.set_index('date', inplace=True)
weather.head()
```

```python
print(weather.isnull().sum())
weather.drop(['location', 'loc_name', 'rain'], axis=1, inplace=True)
```

```python
print(weather.describe())
print(weather.corr())
```

```python
import matplotlib.pyplot as plt

weather.hist(bins=30, figsize=(11.326, 7))
plt.show()
```

```python
# 전운량과 일조량의 관계
x = weather.cloud
y = weather.sunshine
plt.scatter(x,y, color='green', alpha=0.5)
plt.title("cloud-sunshine scatter plot")
plt.xlabel("clouds")
plt.ylabel("sunshine")
plt.show()
```

```python
# 전운량과 습도 사이의 관계
x = weather.cloud
y = weather.humi
plt.scatter(x,y, color='blue', alpha=0.5)
plt.title("cloud-humi scatter plot")
plt.xlabel("clouds")
plt.ylabel("humidity")
plt.show()
```
