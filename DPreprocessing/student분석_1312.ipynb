```python
from itertools import groupby
from signal import pthread_sigmask

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("./assets/student.csv")
df.head()
```

```python
# 히스토그램
plt.hist(df.StudyHours,color='blue')
plt.xlabel("Study Hours")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.scatter(
    x=df.StudyHours,
    y=df.AssignmentScore,
    color='black',
    s=300,
    marker='*')
plt.ylabel("Assignment Score")
plt.xlabel("Study Hours")
plt.show()

# 일반 Pandas 라이브러리를 통헤서도 표현 가능
df.plot(kind='scatter',x='StudyHours',y='AssignmentScore')

```

```python
# 통과와 미통과 집단별 평균 구한 후, 공부시간 비교
result_mean = df.groupby('Result')['StudyHours'].mean()
result_mean
```

```python
# 두 집단을 막대그래프로 나타내기
plt.bar(result_mean.index,result_mean.values)
plt.xticks([0, 1])
plt.xlabel('Result')
plt.ylabel('Average Study Hours')
plt.title('Average Study Hours by Result')
plt.show()
```

```python
# 출석률과 과제점수 간의 관계 파악
plt.scatter(x=df.Attendance,y=df.AssignmentScore,color='black',s=100,marker='o')
plt.xlabel("Attendance")
plt.ylabel("Assignment Score")
plt.show()
# 출석률이 높을수록 과제점수가 높은 양의 상관관계를 확인할 수 있음
```

```python
gender_mean = df.groupby('Gender')["Result"].mean()
print(gender_mean)

plt.bar(gender_mean.index,gender_mean.values)
plt.ylim(0, 1)
plt.xlabel("gender")
plt.ylabel("Percentage of Results")
plt.title('Percentage of Results by Gender')
plt.show()
```
