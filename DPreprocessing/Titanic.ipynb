```python
import pandas as pd

titanic = pd.read_csv('./assets/train.csv')
titanic.head()

for i in titanic.columns:
    print(i)
print(titanic.isnull().sum())

avg = titanic['Age'].mean()
titanic['Age']=titanic['Age'].fillna(avg, inplace=True)


print(type(titanic['Age']))
```

```python

titanic.drop(['Cabin'],axis=1,inplace=True)
titanic.head()
```

```python
print(titanic['Embarked'].value_counts())
titanic['Embarked'] = titanic['Embarked'].fillna('S')
```

```python
titanic.sort_values(by=['Age', 'Fare'], axis=0, inplace=True, ascending=False)

titanic.head(10)
```
